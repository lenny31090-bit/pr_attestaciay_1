```python
# Сгруппируем данные по оттоку и рассчитаем среднее количество сессий, после выведем график
```

```python
sessions_plot = df.groupby('is_churned')['total_sessions_count'].mean()
```

```python
sessions_plot.plot(kind='bar', color=['skyblue', 'salmon'])
```

```python
orders_plot = df.groupby('is_churned')['total_orders_count'].mean()
orders_plot.plot(kind='bar', color=['lightgreen', 'coral'])
```
